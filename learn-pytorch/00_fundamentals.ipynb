## Matrix multiplication

```python
# main two rules for matrix multiplication
# 1. inner dims must match
# 2. resulting matrix has the shape of the outer dims
tensor = torch.tensor([1, 2, 3])
tensor.shape
```

```python
# element-wise multiplication
# [1*1, 2*2, 3*3] = [1, 4, 9]
tensor * tensor
```

```python
# matrix multiplication
# [1*1 + 2*2 + 3*3] = [14]
tensor @ tensor # OR tensor.matmul(tensor)
```

```python
# transposing
# torch.transpose(input, dim0, dim1) -> input is the desired tensor to transpose. dim0 and dim1 are the dims to be swapped.
# tensor.T where tensor is the desired tensor to transpose.
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B.T)
```

```python
torch.mm(tensor_A, tensor_B.T) # shorthand for tensor.matmul
```

```python
torch.manual_seed(42)
linear = torch.nn.Linear(
    in_features=2, # matches the inner dim of the input
    out_features=6 # matches the outer dim of the output
)

x = tensor_A
output = linear(x) # (3, 2) x (2, 6) => (3, 6)
print(f"{x.shape=}")
print(f"{output.shape=}")
print(output)
```

## Reshaping, stacking, squeezing, and unsqueezing

```python
# Oh boy do I need this section
x = torch.arange(1., 8.)
x, x.shape
```

```python
x_reshaped = x.reshape(1, 7) # add an extra dimension
x_reshaped, x_reshaped.shape # not guaranteed whether or not reshaped shares its data or is a new tensor altogether.
```

```python
# change view
# torch.view creates a view of the original tensor. the new tensor ALWAYS shares its data with the original tensor. if you change the original tensor, the viewed tensor will change and vice versa. like aliasing.
z = x.view(1, 7)
z, z.shape
```

```python
x.fill_(1)
z
```

```python
x_reshaped
```

```python
x = torch.arange(1., 8.)
x_stacked = torch.stack([x, x, x, x], dim=0) # since x.shape is [7], this creates a tensor w shape [4, 7]
x_stacked, x_stacked.shape
```

```python
x = torch.arange(1., 8.)
x_stacked = torch.stack([x, x, x, x], dim=1) # since x.shape is [7], this creates a tensor w shape [7, 4]
x_stacked, x_stacked.shape
```

```python
# removing all single dimensions from a tensor. squeeze the tensor to only have dimensions over 1
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

x_squeezed = x_reshaped.squeeze()

print(f"New tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# to reverse torch.squeeze, you can do torch.unsqueeze to add a dim=1 at a specific index.
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

x_unsqueezed = x_squeezed.unsqueeze(dim=0)

print(f"New tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}") # expecting a tensor of shape [1, 7]
```

```python
# can also rearrange the order of axes values with torch.permute(input, dims)
x_original = torch.rand(size=(224, 224, 3))

x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0.

print(f"Previous shape: {x_original.shape}") # [224, 224, 3]
print(f"New shape: {x_permuted.shape}") # expecting [3, 224, 224]
# permuting shares a VIEW (sharing the same data as the original), so essentially like an alias
```

## Reproducibility

```python
# two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A: {random_tensor_A}")
print(f"Tensor B: {random_tensor_B}")
print(f"Does Tensor A equal Tensor B anywhere? {random_tensor_A == random_tensor_B}")
```

```python
RANDOM_SEED = 60
torch.manual_seed(seed=RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

# have to reset the seed every time a new rand() is called.
torch.random.manual_seed(seed=RANDOM_SEED) # if this line is commented out, then tensor_D is different to tensor_C.
random_tensor_D = torch.rand(3, 4)

print(f"Tensor C: {random_tensor_C}")
print(f"Tensor D: {random_tensor_D}")
print(f"Does Tensor C equal Tensor D anywhere?")
random_tensor_C == random_tensor_D
```

## Putting tensors and models on the GPU
